```python
import pandas as pd
import numpy as np
```

```python

# 1. Load data
df = pd.read_csv('city_temperature.csv')
```

```python
print(df.head())
print(df.describe())
```

```python
# Select two cities
city_a = df[df['City'] == 'Delhi']['AvgTemperature']
city_b = df[df['City'] == 'Algiers']['AvgTemperature']
```

```python
# 2. Range
range_a = city_a.max() - city_a.min()
range_b = city_b.max() - city_b.min()
print('Range - Delhi:', range_a)
print('Range - Algiers:', range_b)
```

```python
# 3. Variance
var_a = city_a.var()
var_b = city_b.var()

print('Variance - Delhi:', var_a)
print('Variance - Algiers:', var_b)
```

```python
# 4. Standard Deviation
std_a = city_a.std()
std_b = city_b.std()

print('Std Dev - Delhi:', std_a)
print('Std Dev - Algiers:', std_b)
```

```python
# 5. Coefficient of Variation
cv_a = (std_a / city_a.mean()) * 100
cv_b = (std_b / city_b.mean()) * 100

print('CV% - Delhi:', cv_a)
print('CV% - Algiers:', cv_b)
```

```python
# 6. Interpretation
if std_a < std_b:
    print('Delhi has more consistent temperatures than Algiers')
else:
    print('Algiers has more consistent temperatures than Delhi')
```
